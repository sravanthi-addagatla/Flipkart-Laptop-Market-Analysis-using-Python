#### 1. Brand Market Presence

- Insight: Brands such as HP, Dell, Lenovo, ASUS, and Acer offer the highest number of laptop models, indicating a stronger presence in the market and a wider range of options for customers.

#### 2. Most Preferred Processor

- Insight: Intel processors dominate the laptop market, making them the preferred choice among manufacturers. AMD processors are the next most common, while Apple processors appear mainly in premium devices.

#### 3. Customer RAM Preference

- Insight: 8 GB RAM is the most common configuration, showing that it is the current market standard for everyday users. 16 GB RAM is increasingly popular among professionals and gamers.

#### 4. Popular Storage Option

- Insight: 512 GB SSD is the most frequently offered storage option because it provides a good balance between storage capacity, speed, and affordability.

#### 5. Premium Laptop Segment

- Insight: Premium laptops (priced at ₹50,000 or above) typically include better processors, larger RAM capacities, larger SSDs, and additional features compared to budget laptops.

#### 6. Brand Pricing Strategy

- Insight: Some brands focus on affordable laptops, while others concentrate on premium products. This shows that manufacturers target different customer segments based on price.

#### 7. Discount Strategy

- Insight: Brands offering higher average discounts are likely using promotional pricing to increase sales and compete more effectively in the market.

#### 8. Customer Ratings

- Insight: Brands with higher average customer ratings generally have better customer satisfaction, making them more attractive to potential buyers.

#### 9. Processor and Price Relationship

- Insight: Laptops with high-end processors generally have higher prices because they deliver better performance, making them suitable for gaming, programming, and professional workloads.

#### 10. Buying Recommendation

- Insight: Customers looking for value for money should compare Price, Discount, RAM, SSD, Processor, and Rating together instead of making a decision based on price alone.
