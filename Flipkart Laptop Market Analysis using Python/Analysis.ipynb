```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("laptops_analysis.csv")
```

```python
df
```

### Analysis

#### 1. Processor Comparison

```python
plt.figure(figsize=(6,4))

df["Processor"].value_counts().plot(kind="bar")

plt.title("Processor Comparison")
plt.xlabel("Processor Company")
plt.ylabel("Number of Laptops")
plt.show()
```

##### Analysis
- Intel processors have the highest number of laptop models.
- AMD processors are the second most common.
- Most of the processors belongs to Intel and AMD but different generations
- Apple processors are available in fewer models compared to Intel and AMD

#### 2. RAM Comparison

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x="RAM")

plt.title("RAM Comparison")
plt.xlabel("RAM (GB)")
plt.ylabel("Number of Laptops")
plt.show()
```

##### Analysis
- 16 GB RAM is the most common configuration.
- 16 GB RAM is widely available in premium laptops.
- 32 GB RAM is found mainly in high-performance gaming or workstation laptops.

#### 3. SSD Comparison

```python
plt.figure(figsize=(6,4))

df["SSD"].value_counts().plot(kind="bar")

plt.title("SSD Comparison")
plt.xlabel("SSD Capacity")
plt.ylabel("Number of Laptops")
plt.xticks(rotation=45)
plt.show()
```

##### Analysis
- 512 GB SSD is the most popular storage option.
- 1 TB SSD is available mainly in premium laptops.
- 256 GB SSD is common in entry-level laptops.

#### 4. Brand Comparison

```python
plt.figure(figsize=(8,6))

df["Brand"].value_counts().plot(kind="bar")

plt.title("Brand Comparison")
plt.xlabel("Laptop Brand")
plt.ylabel("Number of Models")
plt.xticks(rotation=45)
plt.show()
```

#### Analysis
- Brands such as ASUS, HP, Lenovo, Acer, and Dell offer the largest number of laptop models.
- Apple has fewer models but mainly focuses on premium laptops.
- Some brands have only a limited number of products in the dataset.

#### 5. Premium Laptop Analysis

```python
plt.figure(figsize=(6,5))

sns.countplot(data=df, x="Price_Category")

plt.title("Premium vs Budget Laptops")
plt.xlabel("Category")
plt.ylabel("Number of Laptops")
plt.show()
```

##### Analysis
- HP, Lenovo, and Dell provide laptops across both budget and premium price ranges.
- ASUS offers a mix of affordable and gaming laptops.
