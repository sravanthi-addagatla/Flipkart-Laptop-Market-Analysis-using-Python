```python
laptop_brand=[]
Processor=[]
RAM=[]
SSD=[]
Display=[]
price=[]
discount=[]
rating=[]
```

```python
import requests
from bs4 import BeautifulSoup
import time
for k in range(1,42):
    url=f"https://www.flipkart.com/search?q=laptops&otracker=search&otracker1=search&marketplace=FLIPKART&as-show=on&as=off&page={k}"
    headers = {"User-Agent": "Mozilla/5.0"}
    response = requests.get(url, headers=headers)
    response
    soup=BeautifulSoup(response.text,"html.parser")
    all_laps = soup.select("div.jIjQ8S")
    all_laps
    for i in all_laps:
        processor_val = "NA"
        ram_val = "NA"
        ssd_val = "NA"
        display_val = "NA"
        laptop_brand.append(i.select_one("div.RG5Slk").text)
        
        for j in i.select(".DTBslk"):
            text = j.text
            
            if "Processor" in text and processor_val == "NA":
                processor_val = text

            elif "RAM" in text and ram_val == "NA":
                ram_val = text

            elif "SSD" in text and ssd_val == "NA":
                ssd_val = text

            elif "Display" in text and display_val == "NA":
                display_val = text

        Processor.append(processor_val)
        RAM.append(ram_val)
        SSD.append(ssd_val)
        Display.append(display_val)


    
        price_tag = i.select_one(".hZ3P6w.DeU9vF")
        if price_tag:
            price.append(i.select_one(".hZ3P6w.DeU9vF").text)
        else:
            price.append("Price not available")
    
        discount_tag = i.select_one(".HQe8jr")
        if discount_tag:
            discount.append(i.select_one(".HQe8jr").text)
        else:
            discount.append("None")
            
        rating_tag = i.select_one(".MKiFS6")
        if rating_tag:
            rating.append(i.select_one(".MKiFS6").text)
        else:
            rating.append("None")
            
        print("-" * 50)
    print(f"Page {k} is Scraped Successfully!")
    time.sleep(2)
    print("Lets take a pause as a Responsible Scraper")
    print("*" * 30)
```

```python
print("laptop_brand",len(laptop_brand))
print("Processor",len(Processor))
print("RAM",len(RAM))
print("SSD",len(SSD))
print("Display",len(Display))
print("price",len(price))
print("discount",len(discount))
print("rating",len(rating))
```

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Product Name":laptop_brand, "Processor":Processor, "RAM":RAM, "SSD":SSD, "Display":Display,
    "Price":price, "Discount":discount, "Rating":rating
})

df.to_csv("laptops_analysis.csv", index=False)
print("CSV file saved successfully!")
```
