```python
import pandas as pd
```

```python
df = pd.read_csv("laptops_analysis.csv")
df
```

#### 1. Which laptop brand has the highest number of models?

```python
df["Brand"].value_counts()
```

##### The brand with the highest number of laptop models has the strongest product presence in the market and offers customers more choices.

#### 2. Which processor company is most popular?

```python
df["Processor"].value_counts()
```

##### The processor company with the highest count is the most preferred by laptop manufacturers in the dataset.

#### 3. What is the average price of laptops for each brand?

```python
df.groupby("Brand")["Price"].mean().sort_values(ascending=False)
```

##### Brands with higher average prices mainly target premium customers, while lower-priced brands focus on budget-conscious buyers.

#### 4. Which RAM capacity is most common?

```python
df["RAM"].value_counts()
```

##### The most common RAM configuration indicates the current market standard for laptops.

#### 5. Which SSD storage capacity is most popular?

```python
df["SSD"].value_counts()
```

##### The most frequently available SSD capacity reflects customer demand and manufacturer preferences.

#### 6. Which brands offer the highest average discount?

```python
df.groupby("Brand")["Discount"].mean().round(2).sort_values(ascending=False)
```

##### Brands providing higher average discounts are likely using promotional pricing to attract more customers.

#### 7. How many Budget and Premium laptops are available?

```python
df["Price_Category"].value_counts()
```

##### This helps understand whether the market is dominated by budget laptops or premium laptops.

#### 8. Which processor company has the highest average laptop price?

```python
df.groupby("Processor")["Price"].mean().round(2).sort_values(ascending=False)
```

##### This comparison shows which processor brands are commonly used in premium-priced laptops.

#### 9. Which are the Top 10 most expensive laptops?

```python
df[["Product Name","Brand","Price"]].sort_values(
    by="Price",
    ascending=False
).head(10)
```

##### These laptops represent the premium segment and usually include high-end specifications and advanced features.

#### 10. Which brands have the highest average ratings?

```python
df.groupby("Brand")["Rating"].mean().round(1).sort_values(ascending=False)
```

##### Brands with higher average ratings generally receive better customer satisfaction and positive reviews.
